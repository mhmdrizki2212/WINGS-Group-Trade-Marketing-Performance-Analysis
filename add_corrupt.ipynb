```python
import pandas as pd
import numpy as np
import random
import os


# ==================================================
# SETTING
# ==================================================

random.seed(42)
np.random.seed(42)



INPUT_FOLDER = "clean_data"

OUTPUT_FOLDER = "raw_data"


os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)



# ==================================================
# LOAD DATA
# ==================================================


sales = pd.read_csv(
    f"{INPUT_FOLDER}/fact_sales.csv"
)


outlet = pd.read_csv(
    f"{INPUT_FOLDER}/dim_outlet.csv"
)


promotion = pd.read_csv(
    f"{INPUT_FOLDER}/fact_promotion.csv"
)


campaign = pd.read_csv(
    f"{INPUT_FOLDER}/fact_digital_campaign.csv"
)



print("Data berhasil dibaca")



# ==================================================
# 1. FACT SALES CORRUPTION
# ==================================================


print("Processing sales...")



# -----------------------------------
# A. Duplicate 68 rows
# -----------------------------------


duplicate_rows = sales.sample(
    68,
    random_state=42
)


sales = pd.concat(
    [
        sales,
        duplicate_rows
    ],
    ignore_index=True
)



# -----------------------------------
# B. Missing Value
# -----------------------------------


# discount kosong
idx = sales.sample(
    100,
    random_state=1
).index


sales.loc[
    idx,
    "discount"
] = np.nan



# quantity kosong

idx = sales.sample(
    50,
    random_state=2
).index


sales.loc[
    idx,
    "quantity"
] = np.nan



# outlet kosong

idx = sales.sample(
    20,
    random_state=3
).index


sales.loc[
    idx,
    "outlet_id"
] = np.nan



# -----------------------------------
# C. SKU Typo
# -----------------------------------


idx = sales.sample(
    100,
    random_state=4
).index



sales.loc[
    idx,
    "sku_id"
] = (

    sales.loc[
        idx,
        "sku_id"
    ]

    .astype(str)

    .str.replace(
        "-",
        "",
        n=1
    )

)



# -----------------------------------
# D. Numeric format issue
# -----------------------------------


idx = sales.sample(
    200,
    random_state=5
).index



sales.loc[
    idx,
    "gross_sales"
] = (

    sales.loc[
        idx,
        "gross_sales"
    ]

    .apply(
        lambda x:
        f"{int(x):,}"
    )

)



# Save

sales.to_csv(

    f"{OUTPUT_FOLDER}/fact_sales_raw.csv",

    index=False

)



# ==================================================
# 2. OUTLET CORRUPTION
# ==================================================


print("Processing outlet...")



idx = outlet.sample(
    20,
    random_state=10
).index



outlet.loc[
    idx,
    "outlet_name"
] = (

    outlet.loc[
        idx,
        "outlet_name"
    ]

    .str.lower()

)



# tambah spasi

idx = outlet.sample(
    10,
    random_state=11
).index


outlet.loc[
    idx,
    "outlet_name"
] = (

    outlet.loc[
        idx,
        "outlet_name"
    ]

    + " "

)



outlet.to_csv(

    f"{OUTPUT_FOLDER}/dim_outlet_raw.csv",

    index=False

)



# ==================================================
# 3. PROMOTION CORRUPTION
# ==================================================


print("Processing promotion...")



# missing discount

idx = promotion.sample(
    50,
    random_state=20
).index



promotion.loc[
    idx,
    "discount_percentage"
] = np.nan



# missing budget

idx = promotion.sample(
    30,
    random_state=21
).index



promotion.loc[
    idx,
    "promo_budget"
] = np.nan



promotion.to_csv(

    f"{OUTPUT_FOLDER}/fact_promotion_raw.csv",

    index=False

)



# ==================================================
# 4. DIGITAL CAMPAIGN CORRUPTION
# ==================================================


print("Processing campaign...")



# clicks kosong

idx = campaign.sample(
    50,
    random_state=30
).index



campaign.loc[
    idx,
    "clicks"
] = np.nan



# CTR kosong

idx = campaign.sample(
    50,
    random_state=31
).index



campaign.loc[
    idx,
    "ctr"
] = np.nan



campaign.to_csv(

    f"{OUTPUT_FOLDER}/fact_digital_campaign_raw.csv",

    index=False

)



print("==============================")
print("DATA CORRUPTION SELESAI")
print("==============================")
```
