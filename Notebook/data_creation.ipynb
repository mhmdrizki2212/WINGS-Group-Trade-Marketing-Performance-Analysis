```python
# =====================================================
# FACT PROMOTION GENERATOR
# WINGS GROUP TRADE MARKETING ANALYTICS
#
# Target : ~80.000 promotion rows
# Period : 2023 - 2026
# =====================================================


import pandas as pd
import numpy as np
import random
from datetime import timedelta


# =====================================================
# RANDOM SEED
# =====================================================

random.seed(42)
np.random.seed(42)



# =====================================================
# LOAD MASTER DATA
# =====================================================


product = pd.read_csv(
    "dim_product.csv"
)


outlet = pd.read_csv(
    "dim_outlet.csv"
)


date = pd.read_csv(
    "dim_date.csv"
)



# =====================================================
# PARAMETER
# =====================================================


TOTAL_PROMOTION = 80000



# =====================================================
# SKU PROMOTION PROBABILITY
# =====================================================


category_probability = {

    "Food":0.80,

    "Beverage":0.75,

    "Fabric Care":0.70,

    "Snack":0.50,

    "Home Care":0.50,

    "Personal Care":0.40,

    "Baby Care":0.30

}



product["promo_probability"] = (
    product["category"]
    .map(category_probability)
)



# =====================================================
# PROMO TYPE BASED ON CHANNEL
# =====================================================


promo_rules = {


    "Modern Trade":[

        "Price Discount",
        "Bundle Promo",
        "Display Promo",
        "Sampling"

    ],


    "Wholesale":[

        "Trade Discount",
        "Bulk Discount",
        "Carton Promo"

    ],


    "E-Commerce":[

        "Flash Sale",
        "Voucher",
        "Cashback",
        "Free Shipping",
        "Campaign Event"

    ]

}



# =====================================================
# DISCOUNT RULE
# =====================================================


def generate_discount(channel):


    if channel == "Modern Trade":

        return random.randint(
            5,
            20
        )


    elif channel == "Wholesale":

        return random.randint(
            10,
            30
        )


    else:

        return random.randint(
            15,
            50
        )



# =====================================================
# PROMO DURATION
# =====================================================


def generate_duration(promo_type):


    if promo_type == "Flash Sale":

        return random.randint(
            1,
            3
        )


    elif promo_type in [
        "Voucher",
        "Cashback"
    ]:

        return random.randint(
            3,
            14
        )


    elif promo_type in [
        "Campaign Event",
        "Display Promo"
    ]:

        return random.randint(
            14,
            45
        )


    else:

        return random.randint(
            14,
            30
        )



# =====================================================
# PROMO BUDGET
# =====================================================


def generate_budget(channel):


    if channel == "Modern Trade":

        return random.randint(
            5000000,
            100000000
        )


    elif channel == "Wholesale":

        return random.randint(
            10000000,
            250000000
        )


    else:

        return random.randint(
            5000000,
            500000000
        )



# =====================================================
# GENERATE PROMOTION
# =====================================================


promotion=[]



promo_counter=1



while len(promotion) < TOTAL_PROMOTION:


    # -------------------------
    # Select Product
    # -------------------------

    sku = product.sample(
        1,
        weights="promo_probability"
    ).iloc[0]



    # -------------------------
    # Select Outlet
    # -------------------------

    store = outlet.sample(
        1
    ).iloc[0]



    channel = store["channel"]



    # -------------------------
    # Promo Type
    # -------------------------

    promo_type = random.choice(
        promo_rules[channel]
    )



    # -------------------------
    # Date
    # -------------------------

    start = pd.to_datetime(
        date.sample(1)["date"].values[0]
    )


    duration = generate_duration(
        promo_type
    )


    end = start + timedelta(
        days=duration
    )


    # limit sampai 2026
    if end.year > 2026:
        continue



    # -------------------------
    # Seasonal Boost
    # -------------------------

    if start.month in [3,4]:

        if sku["category"] in [
            "Food",
            "Beverage",
            "Fabric Care"
        ]:

            promo_type = (
                "Ramadan Campaign"
            )



    # -------------------------
    # Create Row
    # -------------------------


    promotion.append(

        [

            f"PROMO{promo_counter:06d}",

            sku["sku_id"],

            store["outlet_id"],

            promo_type,

            start.date(),

            end.date(),

            generate_discount(
                channel
            ),

            generate_budget(
                channel
            )

        ]

    )


    promo_counter += 1





# =====================================================
# CREATE DATAFRAME
# =====================================================


fact_promotion = pd.DataFrame(

    promotion,

    columns=[

        "promotion_id",

        "sku_id",

        "outlet_id",

        "promo_type",

        "start_date",

        "end_date",

        "discount_percentage",

        "promo_budget"

    ]

)



# =====================================================
# VALIDATION
# =====================================================


print(
    fact_promotion.head()
)


print("\nTotal Promotion:")
print(
    len(fact_promotion)
)



print("\nPromo Type:")
print(
    fact_promotion["promo_type"]
    .value_counts()
)



# =====================================================
# EXPORT
# =====================================================


fact_promotion.to_csv(

    "fact_promotion_wings.csv",

    index=False

)



print(
    "\nFile berhasil dibuat:"
)

print(
    "fact_promotion_wings.csv"
)
```

```python
# ======================================================
# FACT SALES GENERATOR OPTIMIZED
# WINGS GROUP TRADE MARKETING ANALYTICS
#
# INPUT:
# dim_product.csv
# dim_outlet.csv
# dim_date.csv
# fact_promotion.csv
#
# OUTPUT:
# fact_sales.csv
#
# ROW TARGET:
# 1,435,389
# ======================================================


import pandas as pd
import numpy as np
import random
from tqdm import tqdm



# ======================================================
# RANDOM SEED
# ======================================================

random.seed(42)
np.random.seed(42)



# ======================================================
# LOAD MASTER DATA
# ======================================================


product = pd.read_csv(
    "dim_product.csv"
)


outlet = pd.read_csv(
    "dim_outlet.csv"
)


date = pd.read_csv(
    "dim_date.csv"
)


promotion = pd.read_csv(
    "fact_promotion.csv"
)



print("MASTER DATA LOADED")

print(
    "Product:",
    product.shape
)

print(
    "Outlet:",
    outlet.shape
)

print(
    "Date:",
    date.shape
)

print(
    "Promotion:",
    promotion.shape
)



# ======================================================
# PREP PRODUCT
# ======================================================


category_weight = {

    "Food":0.85,

    "Beverage":0.80,

    "Fabric Care":0.70,

    "Snack":0.65,

    "Home Care":0.60,

    "Personal Care":0.55,

    "Baby Care":0.45

}


product["sales_weight"] = (

    product["category"]

    .map(category_weight)

    .fillna(0.5)

)



# ======================================================
# PREP OUTLET
# ======================================================


tier_multiplier = {

    "A":2.0,

    "B":1.0,

    "C":0.5

}



outlet["sales_multiplier"] = (

    outlet["tier"]

    .map(tier_multiplier)

    .fillna(1)

)



# ======================================================
# PREP DATE
# ======================================================


date["date"] = pd.to_datetime(
    date["date"]
)



# ======================================================
# CREATE PROMO LOOKUP
# ======================================================

print("Creating promotion lookup...")

promotion["start_date"] = pd.to_datetime(
    promotion["start_date"]
)


promotion["end_date"] = pd.to_datetime(
    promotion["end_date"]
)



promo_lookup = {}



for _, row in promotion.iterrows():


    promo_dates = pd.date_range(

        start=row["start_date"],

        end=row["end_date"],

        freq="D"

    )


    for d in promo_dates:


        key = (

            row["sku_id"],

            row["outlet_id"],

            d.strftime("%Y-%m-%d")

        )


        promo_lookup[key] = (

            row["discount_percentage"]

        )



print(
    "Promo lookup size:",
    len(promo_lookup)
)



# ======================================================
# PARAMETER
# ======================================================


TOTAL_ROWS = 1435389


CHUNK_SIZE = 100000



# ======================================================
# GENERATE FUNCTION
# ======================================================


def generate_sales(batch, start_id):


    # ----------------------------
    # Sampling
    # ----------------------------


    date_sample = date.sample(

        batch,

        replace=True

    )


    outlet_sample = outlet.sample(

        batch,

        replace=True

    )


    product_sample = product.sample(

        batch,

        replace=True,

        weights=product["sales_weight"]

    )



    df = pd.DataFrame()



    # ----------------------------
    # Transaction ID
    # ----------------------------


    df["transaction_id"] = [

        f"TRX{i:09d}"

        for i in range(

            start_id,

            start_id + batch

        )

    ]



    # ----------------------------
    # Keys
    # ----------------------------


    df["date_id"] = (

        date_sample["date_id"]

        .values

    )


    df["outlet_id"] = (

        outlet_sample["outlet_id"]

        .values

    )


    df["sku_id"] = (

        product_sample["sku_id"]

        .values

    )



    # ----------------------------
    # Base Quantity
    # ----------------------------


    quantity = []


    for channel, multiplier in zip(

        outlet_sample["channel"],

        outlet_sample["sales_multiplier"]

    ):


        if channel == "Modern Trade":

            qty = random.randint(
                10,
                150
            )


        elif channel == "Wholesale":

            qty = random.randint(
                50,
                800
            )


        else:

            qty = random.randint(
                1,
                20
            )



        quantity.append(

            int(qty * multiplier)

        )



    df["quantity"] = quantity



    # ----------------------------
    # Promo Checking
    # ----------------------------


    discounts = []

    promo_multiplier = []



    for sku, outlet_id, trx_date in zip(

        df["sku_id"],

        df["outlet_id"],

        date_sample["date"]

    ):


        key = (

            sku,

            outlet_id,

            trx_date.strftime("%Y-%m-%d")

        )


        if key in promo_lookup:


            discounts.append(

                promo_lookup[key]

            )


            promo_multiplier.append(

                random.uniform(
                    1.3,
                    2.0
                )

            )


        else:


            discounts.append(

                random.randint(
                    0,
                    5
                )

            )


            promo_multiplier.append(

                1

            )



    df["quantity"] = (

        df["quantity"]

        *

        promo_multiplier

    ).astype(int)



    df["discount_percentage"] = discounts



    # ----------------------------
    # Seasonal Effect
    # ----------------------------


    months = (

        date_sample["date"]

        .dt.month

    )


    seasonal = np.where(

        months.isin([3,4]),

        1.3,

        1

    )


    df["quantity"] = (

        df["quantity"]

        *

        seasonal

    ).astype(int)



    # ----------------------------
    # Sales Calculation
    # ----------------------------


    price = (

        product_sample["unit_price"]

        .values

    )



    df["gross_sales"] = (

        df["quantity"]

        *

        price

    )



    df["discount"] = (

        df["gross_sales"]

        *

        df["discount_percentage"]

        /

        100

    ).astype(int)



    df["net_sales"] = (

        df["gross_sales"]

        -

        df["discount"]

    )



    df.drop(

        columns=[

            "discount_percentage"

        ],

        inplace=True

    )


    return df



# ======================================================
# EXPORT
# ======================================================


output_file = "fact_sales.csv"


generated = 0


first_write = True



with tqdm(

    total=TOTAL_ROWS

) as bar:



    while generated < TOTAL_ROWS:



        batch = min(

            CHUNK_SIZE,

            TOTAL_ROWS - generated

        )



        sales = generate_sales(

            batch,

            generated + 1

        )



        sales.to_csv(

            output_file,

            mode="a",

            header=first_write,

            index=False

        )



        first_write = False


        generated += batch


        bar.update(batch)



print("==============================")
print("FACT SALES SELESAI")
print("TOTAL ROW:", generated)
print("FILE:", output_file)
print("==============================")
```

```python
# =====================================================
# FACT DIGITAL CAMPAIGN GENERATOR
# WINGS GROUP TRADE MARKETING ANALYTICS
#
# Rows : 50,000
#
# =====================================================


import pandas as pd
import numpy as np
import random
from datetime import timedelta



# =====================================================
# RANDOM SEED
# =====================================================

random.seed(42)
np.random.seed(42)



# =====================================================
# LOAD MASTER
# =====================================================


product = pd.read_csv(
    "dim_product.csv"
)


date = pd.read_csv(
    "dim_date.csv"
)



# =====================================================
# PARAMETER
# =====================================================


TOTAL_CAMPAIGN = 50000



# =====================================================
# PRODUCT WEIGHT
# =====================================================


category_weight = {


    "Food":0.90,

    "Beverage":0.85,

    "Fabric Care":0.80,

    "Snack":0.70,

    "Home Care":0.60,

    "Personal Care":0.50,

    "Baby Care":0.40

}



product["campaign_weight"] = (

    product["category"]

    .map(category_weight)

    .fillna(0.5)

)



# =====================================================
# PLATFORM
# =====================================================


platform_rule = {


    "TikTok":25,

    "Shopee Ads":20,

    "Instagram":15,

    "Facebook":10,

    "Google Ads":10,

    "YouTube":10,

    "Tokopedia Ads":10

}



platforms=list(
    platform_rule.keys()
)


platform_weights=list(
    platform_rule.values()
)



# =====================================================
# CAMPAIGN TYPE
# =====================================================


campaign_types=[

    "Brand Awareness",

    "Product Launch",

    "Promo Campaign",

    "Ramadan Campaign",

    "11.11 Campaign",

    "12.12 Campaign",

    "Influencer Campaign",

    "Flash Sale Campaign"

]



# =====================================================
# GENERATE
# =====================================================


campaign=[]



for i in range(
    TOTAL_CAMPAIGN
):


    # Product

    sku = product.sample(

        1,

        weights=product["campaign_weight"]

    ).iloc[0]



    # Platform

    platform=random.choices(

        platforms,

        weights=platform_weights

    )[0]



    # Campaign Type

    campaign_type=random.choice(

        campaign_types

    )



    # Date

    start=pd.to_datetime(

        date.sample(1)["date"].values[0]

    )



    # Duration

    if campaign_type=="Flash Sale Campaign":

        duration=random.randint(
            1,
            7
        )


    elif campaign_type=="Brand Awareness":

        duration=random.randint(
            30,
            90
        )


    else:

        duration=random.randint(
            7,
            60
        )



    end=start+timedelta(
        days=duration
    )



    if end.year > 2026:

        continue



    # Spend

    spend=random.randint(

        5000000,

        1000000000

    )



    # Impression

    impressions=random.randint(

        10000,

        20000000

    )



    # CTR

    if platform in [

        "Shopee Ads",

        "Tokopedia Ads",

        "Google Ads"

    ]:


        ctr=random.uniform(

            0.02,

            0.08

        )


    else:


        ctr=random.uniform(

            0.005,

            0.03

        )



    clicks=int(

        impressions*ctr

    )



    # Engagement

    engagement=int(

        impressions*

        random.uniform(

            0.01,

            0.10

        )

    )



    # Conversion

    conversion=random.uniform(

        0.01,

        0.08

    )



    campaign.append([


        f"CMP{i+1:06d}",


        sku["sku_id"],


        platform,


        campaign_type,


        start.date(),


        end.date(),


        spend,


        impressions,


        clicks,


        engagement,


        round(

            ctr*100,

            2

        ),


        round(

            conversion*100,

            2

        )


    ])




# =====================================================
# DATAFRAME
# =====================================================


fact_campaign=pd.DataFrame(

    campaign,

    columns=[

        "campaign_id",

        "sku_id",

        "platform",

        "campaign_type",

        "start_date",

        "end_date",

        "spend",

        "impressions",

        "clicks",

        "engagement",

        "ctr",

        "conversion_rate"

    ]

)



# =====================================================
# VALIDATION
# =====================================================


print(
    fact_campaign.head()
)


print(
    "\nTotal Campaign:",
    len(fact_campaign)
)



print(

    "\nPlatform Distribution:"

)


print(

    fact_campaign["platform"]

    .value_counts()

)



# =====================================================
# EXPORT
# =====================================================


fact_campaign.to_csv(

    "fact_digital_campaign.csv",

    index=False

)



print(

    "\nFile berhasil dibuat: fact_digital_campaign.csv"

)
```
