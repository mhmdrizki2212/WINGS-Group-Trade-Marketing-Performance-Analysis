```python
# SKU yang ada di fact_sales tetapi tidak ada di dim_product

missing_sku_before = set(
    fact_sales["sku_id"]
) - set(
    dim_product["sku_id"]
)

print("Jumlah SKU tidak match:", len(missing_sku_before))

print(list(missing_sku_before)[:20])
```

```python
# Membuat mapping dari dim_product

sku_mapping = {
    sku.replace("-", ""): sku
    for sku in dim_product["sku_id"]
}
```

```python
list(sku_mapping.items())[:10]
```

```python
fact_sales["sku_id_clean"] = (
    fact_sales["sku_id"]
    .str.replace("-", "", regex=False)
    .map(sku_mapping)
)
```

```python
mapped_count = fact_sales["sku_id_clean"].notnull().sum()

print(
    "SKU berhasil dimapping:",
    mapped_count
)
```

```python
unmapped = fact_sales[
    fact_sales["sku_id_clean"].isnull()
]

print(
    "SKU gagal dimapping:",
    len(unmapped)
)
```

```python
unmapped["sku_id"].unique()
```

```python
fact_sales["sku_id"] = fact_sales["sku_id_clean"]
```

```python
fact_sales.drop(
    columns=["sku_id_clean"],
    inplace=True
)
```

```python
missing_sku_after = set(
    fact_sales["sku_id"]
) - set(
    dim_product["sku_id"]
)

print(
    "Jumlah SKU tidak match setelah cleaning:",
    len(missing_sku_after)
)
```

```python
# Check duplicate transaction_id

duplicate_before = fact_sales[
    fact_sales["transaction_id"].duplicated()
]

print("Jumlah duplicate:", len(duplicate_before))
```

```python
# Remove duplicate berdasarkan transaction_id

fact_sales_clean = fact_sales.drop_duplicates(
    subset=["transaction_id"],
    keep="first"
)
```

```python
duplicate_after = fact_sales_clean[
    fact_sales_clean["transaction_id"].duplicated()
]

print("Duplicate setelah cleaning:", len(duplicate_after))
```

```python
print("Jumlah data sebelum:", len(fact_sales))
print("Jumlah data sesudah:", len(fact_sales_clean))
print("Data yang dihapus:", len(fact_sales)-len(fact_sales_clean))
```
